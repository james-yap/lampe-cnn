```python
import numpy as np

from shared.mat_reader import MatReader

mat_reader = MatReader("/data/lampe/dataset")

last_idx = np.where(mat_reader.class_labels == 0)[0][-1]

a = mat_reader.fov_ids[:last_idx + 1]
b = mat_reader.fov_ids[last_idx + 1:]

com = np.intersect1d(a, b)

overlapping_idx = np.where(np.isin(mat_reader.fov_ids, com))

for idx in overlapping_idx[0]:
    print(idx, mat_reader.fov_ids[idx], mat_reader.class_labels[idx])
```
