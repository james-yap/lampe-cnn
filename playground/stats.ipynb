## Box Plot

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 1. Put your data into a Pandas DataFrame format
data = pd.DataFrame(
    {
        "Model": ["ResNet18 + FC 1"] * 10 + ["MicroNet + SVM"] * 10,
        "Accuracy": [0.9333, 0.5333, 0.7857, 0.5000, 0.8750, 0.8750, 0.6875, 0.8000, 0.6250, 1.0000]  # 0.76
        + [0.9375, 0.8, 0.9333, 0.8667, 0.7333, 0.5333, 0.8, 0.6875, 0.6875, 0.6875],
    }
)

plt.figure(figsize=(3, 4))

# Box
sns.boxplot(x="Model", y="Accuracy", data=data, palette="Set2", width=0.1)

# Dots
sns.stripplot(
    x="Model", y="Accuracy", data=data, color="black", alpha=1.0, jitter=True, size=3
)

plt.ylabel("Accuracy", fontsize=12)
plt.xlabel("Hyperparameters", fontsize=12)
plt.title("Model Performance Comparison", fontsize=14)
plt.tight_layout()

# plt.savefig("advanced_boxplot.png")
plt.show()
```

## Stats Test

```python
from scipy import stats

# Example data: 10 accuracy scores per configuration
baseline = [0.877, 0.868, 0.879, 0.892, 0.866, 0.866, 0.893, 0.881, 0.863, 0.878]
tech_a =   [0.866, 0.878, 0.888, 0.852, 0.866, 0.862, 0.854, 0.908, 0.870, 0.884]
tech_b =   [0.916, 0.925, 0.915, 0.932, 0.908, 0.916, 0.914, 0.923, 0.923, 0.911]

# Run the Mann-Whitney U test
mw_stat, mw_p = stats.mannwhitneyu(baseline, tech_a, alternative='two-sided')

print(f"P-value: {mw_p:.4f}")

# Run the Kruskal-Wallis test
kw_stat, kw_p = stats.kruskal(baseline, tech_a, tech_b)

print(f"P-value: {kw_p:.4f}")
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 1. Put your data into a Pandas DataFrame format
data = pd.DataFrame(
    {
        "Model": ["Baseline"] * 10 + ["Tech A"] * 10 + ["Tech B"] * 10,
        "Accuracy": baseline + tech_a + tech_b,
    }
)

plt.figure(figsize=(3, 4))

# Box
sns.boxplot(x="Model", y="Accuracy", data=data, palette="Set2", width=0.1)

# Dots
# sns.stripplot(
#     x="Model", y="Accuracy", data=data, color="black", alpha=1.0, jitter=True, size=3
# )

plt.ylabel("Accuracy", fontsize=12)
plt.xlabel("Hyperparameters", fontsize=12)
plt.title("Model Performance Comparison", fontsize=14)
plt.tight_layout()

# plt.savefig("advanced_boxplot.png")
plt.show()
```
